```python
x=df.iloc[:,:-1]
x
```

```python
y=df.iloc[:,-1]
y
```

```python
x_train,x_test,y_train,y_test=train_test_split(x,y,test_size=0.2,random_state=42)
```

```python
model=LinearRegression()
```

```python
model.fit(x_train, y_train)
y_train_pred=model.predict(x_train)
y_test_pred=model.predict(x_test)
print("Predections generated for both training and testing datasets")
```

```python
mse_train=mean_squared_error(y_train,y_train_pred)
r2_train=r2_score(y_train,y_train_pred)
print("MSE for training dataset:",mse_train)
print("R2 score for training dataset:",r2_train)
```

```python
model_slope=model.coef_
print("Model slope:",model_slope)
```

```python
model_intercept=model.intercept_
print("Model intercept:",model_intercept)
```
